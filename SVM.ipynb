```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, f1_score, recall_score
```

```python
df = pd.read_csv(r"C:\Users\DELL\diabetes.csv")
```

```python
df["Outcome"].value_counts(normalize=True)
```

```python
K = 5
```

```python
X = df.drop(columns=["Outcome"])
y = df["Outcome"]
```

```python
folds = np.random.choice(K, size=df.shape[0], replace = True)
```

```python
X_train = X[folds != 0]
y_train = y[folds != 0]
X_test = X[folds ==0]
y_test = y[folds==0]
```

```python
clf = SVC()
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
cf = confusion_matrix(y_test, y_pred)
cf
```

```python
acc = accuracy_score(y_test, y_pred)
acc
```

```python
prec = precision_score(y_test, y_pred)
prec
```

```python
f1 = f1_score(y_test, y_pred)
f1
```

```python
clf = SVC(kernel = "linear", C=10)
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(y_pred)
```

```python
cf = confusion_matrix(y_test, y_pred)
cf
```

```python
acc = accuracy_score(y_test, y_pred)
acc
```

```python
recall = recall_score(y_test, y_pred)
recall
```

```python
f1 = f1_score(y_test, y_pred)
f1
```

```python
prec = precision_score(y_test, y_pred)
prec
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.svm import SVC
```

```python
df = load_iris()
```

```python
X = df.data[:, :2]
```

```python
y = df.target
```

```python
names = df.target_names[y]
```

```python
mask = (names == "setosa") | (names=="versicolor")
```

```python
X, y = X[mask], y[mask]
```

```python
m1 = SVC(kernel="linear", C=0.1).fit(X,y)
m2 = SVC(kernel="linear", C=100).fit(X,y)
```

```python
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Iris SVM (C=0.1 vs C=100)")
plt.show()
```
